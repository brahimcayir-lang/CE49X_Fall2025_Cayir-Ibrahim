## Exercise 1: Energy Consumption Basics (20 points)

### Part A: Daily Energy Calculation (8 points)

A commercial building has the following daily energy consumption:
- Lighting: 450 kWh
- HVAC: 1200 kWh
- Equipment: 350 kWh
- Other: 180 kWh

**Tasks:**
1. Create variables for each energy component
2. Calculate total daily energy consumption
3. Calculate monthly consumption (30 days)
4. Print results with f-strings

```python
# Exercise 1A: Daily Energy Calculation

# 1️⃣ Create variables for each energy component (in kWh)
lighting = 450       # Lighting energy consumption
hvac = 1200          # Heating, Ventilation, and Air Conditioning system
equipment = 350      # Office or industrial equipment
other = 180          # Miscellaneous energy uses

# 2️⃣ Calculate total daily energy consumption
daily_total = lighting + hvac + equipment + other

# 3️⃣ Calculate total monthly consumption (assuming 30 days)
monthly_total = daily_total * 30

# 4️⃣ Print results using formatted f-strings
# The :.2f format displays numbers with 2 decimal places
print("Daily Energy Consumption:")
print(f"  Lighting: {lighting:.2f} kWh")
print(f"  HVAC: {hvac:.2f} kWh")
print(f"  Equipment: {equipment:.2f} kWh")
print(f"  Other: {other:.2f} kWh")
print(f"  Total: {daily_total:.2f} kWh")

print(f"\nMonthly Consumption: {monthly_total:.2f} kWh")
```

### Part B: Unit Conversion (6 points)

Convert the monthly energy consumption:
- From kWh to MJ (1 kWh = 3.6 MJ)
- From MJ to GJ (1 GJ = 1000 MJ)

```python
# Exercise 1B: Unit Conversion

# Conversion factors (given in the lab instructions)
# 1 kWh = 3.6 MJ
# 1 GJ = 1000 MJ

# 1️⃣ Convert monthly consumption from kWh to MJ
# Formula: Energy (MJ) = Energy (kWh) * 3.6
monthly_mj = monthly_total * 3.6

# 2️⃣ Convert from MJ to GJ
# Formula: Energy (GJ) = Energy (MJ) / 1000
monthly_gj = monthly_mj / 1000

# 3️⃣ Print results neatly using f-strings
print("Monthly Consumption:")
print(f"  {monthly_total:.2f} kWh")
print(f"  {monthly_mj:.2f} MJ")
print(f"  {monthly_gj:.2f} GJ")
```

### Part C: Cost Calculation (6 points)

Calculate costs using standard rate: 0.12 USD/kWh

```python
# Exercise 1C: Cost Calculation

# 1️⃣ Define the standard electricity rate (USD per kWh)
standard_rate = 0.12  # USD/kWh

# 2️⃣ Calculate monthly cost
# Formula: cost = energy (kWh) * rate (USD/kWh)
monthly_cost = monthly_total * standard_rate

# 3️⃣ Calculate annual cost (12 months)
annual_cost = monthly_cost * 12

# 4️⃣ Print results formatted as currency
# , adds thousand separators; .2f shows two decimal places
print("Energy Costs:")
print(f"  Monthly: ${monthly_cost:,.2f}")
print(f"  Annual:  ${annual_cost:,.2f}")
```

## Exercise 2: Building Energy Analysis (25 points)

Analyze energy data for multiple buildings.

### Given Data

```python
# Given data - DO NOT MODIFY
buildings = ['Office A', 'Retail B', 'School C', 'Hospital D', 'Apartment E']
monthly_consumption = [85000, 62000, 48000, 125000, 71000]  # kWh
floor_area = [2500, 1800, 3200, 4000, 2800]  # m^2
```

### Part A: Energy Intensity Calculation (10 points)

Energy Intensity = Monthly Consumption / Floor Area (kWh/m²/month)

```python
# Exercise 2A: Energy Intensity Calculation

# Given data (from the lab instructions)
buildings = ['Office A', 'Retail B', 'School C', 'Hospital D', 'Apartment E']
monthly_consumption = [85000, 62000, 48000, 125000, 71000]  # kWh 
floor_area = [2500, 1800, 3200, 4000, 2800]                  # m²

# 1️⃣ Create an empty list to store calculated energy intensities (kWh/m²/month)
energy_intensity = []

# 2️⃣ Use a for loop with enumerate() to access both index and building name
for i, building in enumerate(buildings):

    # Calculate energy intensity (consumption per floor area)
    # Formula: Energy Intensity = Monthly Consumption / Floor Area
    intensity = monthly_consumption[i] / floor_area[i]

    # Add calculated value to the list
    energy_intensity.append(intensity)

    # Print formatted results
    print(f"{building}: {monthly_consumption[i]} kWh / {floor_area[i]} m² = {intensity:.2f} kWh/m²/month")
```

### Part B: Statistical Analysis (8 points)

Calculate statistics for the building dataset.

```python
# Exercise 2B: Statistical Analysis

# Given data (from previous exercise)
buildings = ['Office A', 'Retail B', 'School C', 'Hospital D', 'Apartment E']
monthly_consumption = [85000, 62000, 48000, 125000, 71000]  # kWh

# 1️⃣ Calculate total monthly consumption across all buildings
# The sum() function adds up all values in the list
total_consumption = sum(monthly_consumption)

# 2️⃣ Calculate average monthly consumption
# Average = total / number of buildings
average_consumption = total_consumption / len(monthly_consumption)

# 3️⃣ Find maximum and minimum consumption values
# max() gives the largest value, min() gives the smallest
max_consumption = max(monthly_consumption)
min_consumption = min(monthly_consumption)

# 4️⃣ Print a formatted summary report
print("=" * 50)
print("Statistical Summary")
print("=" * 50)
print(f"Total Monthly Consumption: {total_consumption:,.0f} kWh")    # formatted with comma separators and no decimals
print(f"Average Monthly Consumption: {average_consumption:,.0f} kWh")
print(f"Maximum Consumption: {max_consumption:,.0f} kWh")
print(f"Minimum Consumption: {min_consumption:,.0f} kWh")
print("=" * 50)

# Format guide:
# :.0f  → 0 decimal places (e.g., 78200)
# :,.0f → 0 decimals + comma separator (e.g., 125,000)
# :.2f  → 2 decimal places (e.g., 78200.45)
# :,.2f → 2 decimals + comma separator (e.g., 125,000.45)

```

### Part C: Find Buildings Above Average (7 points)

Identify buildings with above-average consumption.

```python
# Exercise 2C: Buildings Above Average

# 1️⃣ Create an empty list to store names of buildings with above-average energy use
above_average = []

# 2️⃣ Loop through each building using enumerate()
# enumerate() gives both the index (i) and the building name
for i, building in enumerate(buildings):
    
    # Check if the building's consumption is greater than the overall average
    if monthly_consumption[i] > average_consumption:
        
        # If true, add that building to the "above_average" list
        above_average.append(building)

# 3️⃣ Print results in a clean format
print("\nBuildings with above-average consumption:")
print(above_average)  # prints the list of building names
print(f"Count: {len(above_average)} out of {len(buildings)} buildings")
```

## Exercise 3: Energy Efficiency Classifier (25 points)

Classify buildings based on their annual energy consumption per m².

**Efficiency Ratings:**
- A: < 50 kWh/m²/year (Excellent)
- B: 50-100 kWh/m²/year (Good)
- C: 100-150 kWh/m²/year (Average)
- D: 150-200 kWh/m²/year (Poor)
- F: > 200 kWh/m²/year (Very Poor)

### Part A: Annual Energy Calculation (8 points)

```python
# Exercise 3A: Annual Energy Calculation

# 1️⃣ Create an empty list to store annual intensity values
annual_intensity = []

# 2️⃣ Convert monthly intensity (kWh/m²/month) to annual (kWh/m²/year)
# Formula: Annual Intensity = Monthly Intensity * 12
for intensity in energy_intensity:
    annual = intensity * 12
    annual_intensity.append(annual)

# 3️⃣ Print each building with its annual energy intensity
print("Annual Energy Intensity:")
for i, building in enumerate(buildings):
    print(f"{building}: {annual_intensity[i]:.2f} kWh/m²/year")
```

### Part B: Efficiency Classification (12 points)

```python
# Exercise 3B: Efficiency Classification

# 1️⃣ Create an empty list to store efficiency ratings (A–F)
ratings = []

# 2️⃣ Loop through each building and assign a rating based on annual intensity
for i, building in enumerate(buildings):
    intensity = annual_intensity[i]

    # Classification rules (from lab instructions)
    # A: < 50
    # B: 50–100
    # C: 100–150
    # D: 150–200
    # F: > 200

    if intensity < 50:
        rating = 'A'
    elif intensity < 100:
        rating = 'B'
    elif intensity < 150:
        rating = 'C'
    elif intensity < 200:
        rating = 'D'
    else:
        rating = 'F'

    # Add the assigned rating to the list
    ratings.append(rating)

# 3️⃣ Print formatted efficiency report
print("\n" + "=" * 60)
print("Energy Efficiency Report")
print("=" * 60)
for i, building in enumerate(buildings):
    print(f"{building}: {annual_intensity[i]:.2f} kWh/m²/year - Rating: {ratings[i]}")
```

### Part C: Rating Summary (5 points)

```python
# Exercise 3C: Rating Summary

# 1️⃣ Count how many buildings received each rating
count_a = ratings.count('A')
count_b = ratings.count('B')
count_c = ratings.count('C')
count_d = ratings.count('D')
count_f = ratings.count('F')

# 2️⃣ Print formatted summary
print("\nRating Summary:")
print(f"  A (Excellent): {count_a}")
print(f"  B (Good): {count_b}")
print(f"  C (Average): {count_c}")
print(f"  D (Poor): {count_d}")
print(f"  F (Very Poor): {count_f}")

# 3️⃣ Find which rating is most common
# Create a small dictionary to compare category counts
rating_counts = {
    'A': count_a,
    'B': count_b,
    'C': count_c,
    'D': count_d,
    'F': count_f
}

# Use max() with key argument to find the rating with the highest count
# Normally, max(rating_counts) would compare only the keys alphabetically (e.g., returns 'F' since it's the largest letter)
# For example, if number of D was 10, and number of F was 4, it would gave still F since it is alphabetically the biggest.
# But adding key=rating_counts.get tells Python to compare by the dictionary values instead of the keys
# So it returns the rating (key) that has the highest count value
most_common_rating = max(rating_counts, key=rating_counts.get)

# 4️⃣ Print the most common rating
print(f"\nMost Common Rating: {most_common_rating}")
```

## Exercise 4: Energy Cost Calculator (20 points)

### Part A: Simple Cost Function (8 points)

```python
# Exercise 4A: Simple Cost Function

def calculate_monthly_cost(consumption_kwh, rate_per_kwh):
    """
    Calculate monthly energy cost.
    
    Parameters:
        consumption_kwh (float): Monthly energy consumption in kWh
        rate_per_kwh (float): Energy rate in USD/kWh
    
    Returns:
        float: Monthly cost in USD
    """

    # Formula: Cost = Consumption × Rate
    cost = consumption_kwh * rate_per_kwh
    return cost


# Test the function with example data
test_consumption = 50000  # kWh
test_rate = 0.12          # USD/kWh

# Call the function and store result
result = calculate_monthly_cost(test_consumption, test_rate)

# Print result formatted with commas and 2 decimals
print(f"Monthly cost for {test_consumption} kWh at ${test_rate}/kWh: ${result:,.2f}")
```

### Part B: Peak/Off-Peak Cost Function (12 points)

```python
# Exercise 4B: Peak/Off-Peak Cost Function

def calculate_tiered_cost(total_consumption, peak_percentage=0.6):
    """
    Calculate cost with peak/off-peak pricing.
    
    Parameters:
        total_consumption (float): Total monthly consumption in kWh
        peak_percentage (float): Fraction of consumption during peak hours (default 0.6)
    
    Returns:
        tuple: (peak_cost, off_peak_cost, total_cost)
    """
    peak_rate = 0.15       # USD/kWh (during daytime 08:00–20:00)
    off_peak_rate = 0.08   # USD/kWh (nighttime 20:00–08:00)
    
    # 1️⃣ Calculate how much energy is consumed during each period
    peak_consumption = total_consumption * peak_percentage
    off_peak_consumption = total_consumption * (1 - peak_percentage)
    
    # 2️⃣ Calculate individual costs for each period
    peak_cost = peak_consumption * peak_rate
    off_peak_cost = off_peak_consumption * off_peak_rate
    
    # 3️⃣ Calculate total cost
    total_cost = peak_cost + off_peak_cost
    
    # 4️⃣ Return all three as a tuple
    return peak_cost, off_peak_cost, total_cost


# Test the function
test_consumption = 85000  # kWh
peak_cost, off_peak_cost, total_cost = calculate_tiered_cost(test_consumption, 0.6)

# Display the results neatly
print(f"\nTiered Pricing Analysis for {test_consumption} kWh:")
print(f"  Peak cost (60%): ${peak_cost:,.2f}")
print(f"  Off-peak cost (40%): ${off_peak_cost:,.2f}")
print(f"  Total cost: ${total_cost:,.2f}")
```

### Part A: Most and Least Efficient Buildings (5 points)

```python
# Exercise 5A: Most and Least Efficient

# 1️⃣ Find minimum and maximum annual intensity values
# Lower intensity = more efficient, Higher intensity = less efficient
min_intensity = min(annual_intensity)
max_intensity = max(annual_intensity)

# 2️⃣ Find the index positions of those values in the list
min_index = annual_intensity.index(min_intensity)
max_index = annual_intensity.index(max_intensity)

# 3️⃣ Use the indices to get corresponding building names
most_efficient = buildings[min_index]
least_efficient = buildings[max_index]

# 4️⃣ Print a formatted comparison report
print("=" * 60)
print("Efficiency Comparison")
print("=" * 60)
print(f"Most efficient: {most_efficient}")
print(f"  Intensity: {min_intensity:.2f} kWh/m²/year")
print(f"  Rating: {ratings[min_index]}")
print()
print(f"Least efficient: {least_efficient}")
print(f"  Intensity: {max_intensity:.2f} kWh/m²/year")
print(f"  Rating: {ratings[max_index]}")

# 5️⃣ Calculate percentage difference in intensity
# Formula: ((max - min) / min) × 100
percentage_diff = ((max_intensity - min_intensity) / min_intensity) * 100

print(f"\nPercentage difference: {percentage_diff:.1f}%")
```

### Part B: Energy Savings Potential (5 points)

Calculate savings if all buildings achieved Rating B (100 kWh/m²/year).

```python
# Exercise 5B: Energy Savings Potential

target_intensity = 100  # kWh/m²/year for Rating B
total_savings_kwh = 0

print("\n" + "=" * 60)
print("Energy Savings Potential")
print("=" * 60)
print(f"Target: Rating B ({target_intensity} kWh/m²/year)\n")

# Loop through buildings and calculate savings
for i, building in enumerate(buildings):
    # Consider only buildings worse than B (i.e., annual_intensity > 100)
    if annual_intensity[i] > target_intensity:
        # Current annual consumption (kWh/year) = intensity (kWh/m²/yr) * area (m²)
        current_consumption = annual_intensity[i] * floor_area[i]
        
        # Target annual consumption if it achieved Rating B
        target_consumption = target_intensity * floor_area[i]
        
        # Potential savings (ensure non-negative just in case)
        savings = max(0, current_consumption - target_consumption)
        
        # Accumulate total savings
        total_savings_kwh += savings  # total_savings_kwh = total_savings_kwh + savings
        
        print(f"{building}: Could save {savings:,.0f} kWh/year")

# Annual cost savings at the standard rate (0.12 USD/kWh)
cost_savings = total_savings_kwh * 0.12

print(f"\nTotal Potential Savings: {total_savings_kwh:,.0f} kWh/year")
print(f"Annual Cost Savings: ${cost_savings:,.2f}")
```

## Bonus Challenge (Optional, +5 points)

Create an interactive energy calculator that gets user input.

```python
# ===== Bonus: Interactive Building Energy Calculator (meets all requirements) =====

def calculate_tiered_cost(total_consumption, peak_percentage=0.6):
    """
    Calculate cost with peak/off-peak pricing.
    Peak hours rate: 0.15 USD/kWh (08:00–20:00)
    Off-peak rate:   0.08 USD/kWh (20:00–08:00)
    Returns: (peak_cost, off_peak_cost, total_cost)
    """
    if not 0 <= peak_percentage <= 1:
        raise ValueError("peak_percentage must be between 0 and 1.")
    peak_rate = 0.15 
    off_peak_rate = 0.08
    peak_kwh     = total_consumption * peak_percentage
    off_peak_kwh = total_consumption * (1 - peak_percentage)
    peak_cost     = peak_kwh * peak_rate
    off_peak_cost = off_peak_kwh * off_peak_rate
    return peak_cost, off_peak_cost, peak_cost + off_peak_cost

print("Interactive Building Energy Calculator")
print("=" * 60)

while True:
    # 1) Get building name (or quit)
    name = input("\nEnter building name (or 'quit' to exit): ").strip() # .strip() removes any extra spaces or invisible characters (like tabs/newlines)
    if name.lower() == "quit":                                          # Converts the input to lowercase and checks if the user typed 'quit'
        print("Thank you for using the calculator!")
        break
    if not name:
        print("Error: Building name cannot be empty.")
        continue

    try:
        # 2) Get numeric inputs safely
        monthly_str = input("Enter monthly consumption (kWh): ").strip()
        area_str    = input("Enter floor area (m^2): ").strip()

        monthly_kwh = float(monthly_str)                               # Converts the monthly consumption input (string) into a floating-point number.  
        area_m2     = float(area_str)
        if monthly_kwh <= 0 or area_m2 <= 0:
            raise ValueError("Values must be positive.")               # Manually triggers a ValueError if the user enters zero or negative numbers.

        # 3) Energy intensity and rating
        monthly_intensity = monthly_kwh / area_m2                      # kWh/m^2/month
        annual_intensity  = monthly_intensity * 12                     # kWh/m^2/year

        if   annual_intensity < 50:   rating = "A"
        elif annual_intensity < 100:  rating = "B"
        elif annual_intensity < 150:  rating = "C"
        elif annual_intensity < 200:  rating = "D"
        else:                         rating = "F"

        # 4) Monthly cost using tiered pricing (default 60% peak)
        peak_cost, off_peak_cost, total_cost = calculate_tiered_cost(monthly_kwh, peak_percentage=0.6)

        # 5) Formatted report
        print("\n=== Analysis for", name, "===")
        print(f"Monthly Consumption: {monthly_kwh:,.0f} kWh")
        print(f"Floor Area: {area_m2:,.0f} m^2")
        print(f"Energy Intensity: {annual_intensity:,.2f} kWh/m^2/year")
        print(f"Efficiency Rating: {rating}")
        print(f"Monthly Cost (tiered): ${total_cost:,.2f} "
              f"(Peak: ${peak_cost:,.2f}, Off-peak: ${off_peak_cost:,.2f})")

        # 6) Loop continues for more buildings (per requirement)

    except ValueError:
        print("Error: Please enter valid numeric values (positive numbers).")
    except Exception as e:
        print(f"Unexpected error: {e}")
```
