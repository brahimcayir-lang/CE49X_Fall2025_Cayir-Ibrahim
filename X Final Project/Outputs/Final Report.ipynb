# CE49X Final Project – Civil Engineering & AI Trend Analysis

## 1. Project Overview
This project investigates the integration and adoption rates of Artificial Intelligence (AI) across various Civil Engineering sub-disciplines. By analyzing a curated corpus of news media and academic publications, the study utilizes Natural Language Processing (NLP) to quantify the shift toward digital transformation in the industry. We aim to identify which specific technologies (e.g., Computer Vision, Predictive Analytics) are driving innovation in specific civil sectors (e.g., Structural, Transportation).

## 3. Aim & Objectives
The primary goal is to provide a data-driven assessment of AI's footprint in Civil Engineering.
* **Quantitative Objective:** Measure the frequency of AI-related mentions across sub-disciplines.
* **Qualitative Objective:** Map specific AI technologies to engineering challenges (e.g., AI in site safety).
* **Visualization:** Develop a comprehensive "AI Maturity" heatmap and trend analysis to identify research gaps.

## 4. Project Approach

Our workflow follows a data-driven pipeline:

* **Data Collection  from multiple sources** (RSS feeds, scientific papers, limited scraping, APIs)
* **Data Cleaning & Filtering** (using LLM-assisted methods)
**Text Processing & Analysis**
**Visualization & Interpretation of Trends**

This approach ensures both breadth (industry news) and depth (academic insight), while minimizing noise and bias.

```text
/CE49X Final Project
│
├── /data/
│   └── corpus.db                 # Central SQLite database (Raw & Processed tables)
│
├── /Scripts/
│   ├── task1_scripts.py (Multiple) # Seperate script for each data collection option
│   ├── task2_preprocessing.py    # NLP refinement: Stopword removal & Lemmatization
│   ├── task3_categorization.py   # Discipline tagging & Heatmap logic
│   └── task4_visuals_final.py    # Data Viz: Bar charts, Word Clouds, & Network graphs
│
├── /Outputs/
│   ├── Visuals & Reports         # PNG exports (Visuals) / Statistical reports
│              
│
├── requirements.txt              # Dependency list (pandas, nltk, seaborn, etc.)
└── README.md                     # Project documentation

6. Data Collection Strategy
To ensure a balance between industry real-world application and academic rigor, we utilized a multi-source strategy:

RSS Feeds (Primary Source): This allowed for the capture of 3,000+ articles, providing a "pulse" of the current construction and engineering industry.

Scientific Papers (Supplementary Source): To capture deeper and more reliable trends, we incorporated a limited number of academic papers. While we identified over 3,000, we intentionally used ~200 papers to prevent academic literature from overpowering real-world industry signals.

Web Scraping & APIs: These methods were experimented with but ultimately relegated to minor roles due to high noise levels, structural inconsistencies, and rate-limiting restrictions.

7. Data Cleaning & Refinement
Data quality was prioritized over raw volume. We employed a hybrid cleaning approach:

LLM-Assisted Refinement: Used for initial noise reduction and relevance auditing to ensure only pertinent articles remained in the final corpus.

Rule-Based Cleaning: Aggressive removal of scraper artifacts (e.g., "div", "href", "relnoopener") to ensure that n-gram analysis focused on engineering terms rather than metadata.

Lemmatization: Utilizing the NLTK WordNetLemmatizer to convert words to their root forms (e.g., "buildings" to "building"), ensuring accurate frequency counts across the dataset.

8. Visualization & Analysis
The project utilizes matplotlib, seaborn, and networkx to visualize findings. Key outputs include:

Frequency Distributions: Ranking sub-disciplines by the total volume of AI-related content.

Heatmaps: Illustrating the co-occurrence of specific AI technologies with engineering fields (e.g., Computer Vision usage in Structural Health Monitoring).

Network Graphs: Visualizing the relationship between disparate terms, such as the link between "3D Printing" and "Sustainability."

9. Tools & Technologies
Programming Language: Python 3.10+

Data Handling: SQLite, Pandas, Numpy

NLP & Text Processing: NLTK (WordNetLemmatizer, Punkt), Scikit-Learn (TF-IDF)

Visualization: Matplotlib, Seaborn, NetworkX, WordCloud

Refinement: Custom noise-filtering regex + LLM-assisted metadata cleaning

10. Conclusion & Future Outlook
The synthesized data indicates that Construction Management and Transportation currently lead in AI mentions, driven heavily by site safety monitoring and autonomous infrastructure systems. Structural Engineering is a fast-following sector, primarily focusing on the lifecycle benefits of Digital Twins and SHM.

This analysis provides a baseline for "AI Maturity" in the sector. Future work could include temporal analysis to determine the velocity of AI adoption year-over-year as more generative AI tools enter the construction market.

## Course: CE49X – Introduction to Data Science for Civil Engineering

Institution: Boğaziçi University

Semester: Fall 2025
