# 03 — Analysis
Apply rule-based tagging using CE/AI dictionaries, build co-occurrence matrix, and inspect temporal trends if dates are available.



```python
import pandas as pd
import os

from src.analysis.dictionaries import ce_dict, ai_dict
from src.analysis.classifier import classify_all_articles, cooccurrence_matrix

# Load cleaned data
clean_df = pd.read_csv("data/processed/articles_clean.csv")

# Classify
labeled_df = classify_all_articles(clean_df, ce_dict, ai_dict)
print("Tagged rows:", len(labeled_df))

# Persist enriched data for visualization step
os.makedirs("data/processed", exist_ok=True)
labeled_df.to_csv("data/processed/articles_clean.csv", index=False)

labeled_df.head()[["title", "ce_areas", "ai_techs"]]
```

```python
matrix = cooccurrence_matrix(labeled_df, ce_dict, ai_dict)
matrix
```

```python
# Temporal trend analysis if date column exists
if "date" in labeled_df.columns:
    temp_df = labeled_df.copy()
    temp_df["date"] = pd.to_datetime(temp_df["date"], errors="coerce")
    temp_df = temp_df.dropna(subset=["date"])
    temp_df["year"] = temp_df["date"].dt.year
    trend = temp_df.groupby("year").size()
    display(trend)
else:
    print("No date column found for temporal analysis")
```
