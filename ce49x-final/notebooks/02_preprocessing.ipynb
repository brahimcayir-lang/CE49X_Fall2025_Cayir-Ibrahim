# 02 — Preprocessing
Clean scraped articles (normalize, stopword removal, lemmatize) and produce tokens.

Outputs `data/processed/articles_clean.csv` with `clean_text` and `tokens` columns.



```python
import pandas as pd
from collections import Counter
from nltk.util import ngrams

from src.preprocessing.preprocessing_pipeline import preprocess_all_articles

# Run preprocessing
clean_df = preprocess_all_articles()
print("Rows processed:", len(clean_df))
clean_df.head()

```

```python
# Most frequent words
all_tokens = clean_df["tokens"].explode().dropna()
word_counts = all_tokens.value_counts().head(20)
word_counts

```

```python
# Top 20 bigrams and trigrams
bigrams = Counter(ngrams(all_tokens, 2)).most_common(20)
trigrams = Counter(ngrams(all_tokens, 3)).most_common(20)
print("Top bigrams:", bigrams)
print("Top trigrams:", trigrams)

```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=200, ngram_range=(1, 2))
X = vectorizer.fit_transform(clean_df["clean_text"].fillna(""))
feature_names = vectorizer.get_feature_names_out()

# Mean TF-IDF per term
mean_tfidf = X.mean(axis=0).A1
series_tfidf = pd.Series(mean_tfidf, index=feature_names).sort_values(ascending=False).head(20)
series_tfidf
```
