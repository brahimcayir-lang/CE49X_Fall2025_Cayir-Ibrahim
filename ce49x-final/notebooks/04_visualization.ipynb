# 04 — Visualization
Generate figures: bar charts, heatmap, network graph, and word clouds. Include brief insights after each chart.



```python
import pandas as pd

from src.analysis.dictionaries import ce_dict, ai_dict
from src.analysis.visualization import bar_counts, plot_cooccurrence_heatmap, plot_network, wordclouds_by_area
from src.analysis.classifier import cooccurrence_matrix

labeled_df = pd.read_csv("data/processed/articles_clean.csv")

# If ce_areas/ai_techs saved as lists, ensure proper parsing
for col in ["ce_areas", "ai_techs"]:
    if col in labeled_df.columns and labeled_df[col].dtype == object:
        labeled_df[col] = labeled_df[col].apply(lambda x: eval(x) if isinstance(x, str) and x.startswith('[') else [])

bar_counts(labeled_df.get("ce_areas", []), "Articles by CE Area", "ce_area_counts.png")
bar_counts(labeled_df.get("ai_techs", []), "Articles by AI Tech", "ai_tech_counts.png")

matrix = cooccurrence_matrix(labeled_df, ce_dict, ai_dict)
plot_cooccurrence_heatmap(matrix, "cooccurrence_heatmap.png")
plot_network(matrix, "cooccurrence_network.png")
wordclouds_by_area(labeled_df, ce_dict)

matrix.head()
```

After running the above, review the saved figures in `figures/` and note:
- CE areas with highest counts (dominant in bar chart)
- AI tech popularity
- Strong CE–AI pairs (heatmap/network edges)
- Distinctive vocabulary per CE area (word clouds)
